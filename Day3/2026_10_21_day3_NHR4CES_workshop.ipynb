```python
from dscribe.descriptors import ACSF

acsf = ACSF(
    species=["Si", "O"],
    r_cut=5,
    g2_params=[[1, 1], [1, 2], [1, 3],[1, 4],[1, 2.5]],
    g4_params=[[1, 1, 1], [1, 2, 1], [1, 1, -1], [1, 2, -1]],
    periodic=True
)
```

```python
acsf_functions = acsf.create(atoms)
```

```python
print(acsf_functions[0])

len(acsf_functions[0])
```

```python
from matplotlib import pyplot as plt

for i in range(len(atoms)):
    acsf_values = acsf_functions[i]
    plt.plot(acsf_values, marker='o', linestyle='-')
plt.xlabel('ACSF Feature Index')
plt.ylabel('ACSF Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Check the shape of the output
#print(acsf_functions)
#print(acsf_functions.shape)
n_centers, n_features = acsf_functions.shape
# Plot: Center index vs. the summed feature values for each atom
center_sums = acsf_functions.sum(axis=1)  # Sum of features for each center (atom)
plt.figure(figsize=(6, 4))
plt.plot(range(n_centers), center_sums, 'o-', label='Sum of Features per Center')
plt.xlabel('Center Index (Atom)')
plt.ylabel('Sum of ACSF Features')
plt.title('Sum of ACSF Features per Atom')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from dscribe.descriptors import SOAP
atoms = read("data/training_SiO2.xyz")

soap = SOAP(species=["Si","O"],periodic=True,r_cut=5,n_max=2,l_max=2)
soap_functions = soap.create(atoms)
print(soap_functions.shape)
soap_functions[0]

for i in range(len(soap_functions)):
    soap_values = soap_functions[i]
    plt.plot(soap_values, marker='o', linestyle='-')
plt.xlabel('SOAP Feature Index')
plt.ylabel('SOAP Value')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# We will use ase module to extract the DFT energy from an atom object
# and lern how to calculate the energy using the ML potential

# We import the necesarry modules
from pyace import *
import matplotlib.pyplot as plt

normalize_energy_dict = {
    "O": -0.74034035,
    "Si": -4.14567113,
}

def normalize_energy_struct(energy, symbols, subtract_energy_dict):
    for ele, val in subtract_energy_dict.items():
        n_ele = len(symbols[symbols == ele])
        energy -= n_ele * val
    return energy

Atoms_list = read('data/training_SiO2.xyz', ':')


# randomly pick one structure
atoms_example = Atoms_list[5]

# get the DFT energy that is normalized
e = atoms_example.get_potential_energy()
Energy_example_DFT = normalize_energy_struct(e, atoms_example.symbols, normalize_energy_dict)
print(Energy_example_DFT)


# load the calculator, the ML potential we have trained
calc = PyACECalculator("output_potential.yaml")
# attach the calculator to the atoms object
atoms_example.set_calculator(calc)
Energy_exampl_ML = atoms_example.get_potential_energy()
print(Energy_exampl_ML)




```
